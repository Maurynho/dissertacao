```python
import pandas as pd
dados = pd.read_csv('/content/dados2704.csv', index_col=False)
dados.info()
```

Incluindo a variável  TaxaAtrasoAjustada




```python
dados['TaxaAtrasoAjustada'] = (
    dados['TotalContratacoesAtraso'] /
    (dados['TotalContratacoes'] + 1)
)
```

Incluindo a variável  ComprometimentoParcelaRenda

```python
dados['ParcelaMensalEstimada'] = (
    dados['ValorContratado'] /
    dados['QuantidadeParcelas']
)



dados['ComprometimentoParcelaRenda'] = (
                                        dados['ParcelaMensalEstimada'] /
                                        dados['ReceitaBrutaMensal']
                                    )
```

Incluindo a variável RiscoFinanceiro

```python
dados['RiscoFinanceiro'] = (
    dados['ComprometimentoParcelaRenda'] *
    dados['TaxaAtrasoAjustada']
)
```

```python
dados
```

```python
correlacoes = dados.corr(numeric_only=True)['Alvo'].sort_values(ascending=False)

print(correlacoes)
```

```python
novas_features = [
    'TaxaAtrasoAjustada',
    'ComprometimentoParcelaRenda',
    'RiscoFinanceiro'
]

variaveis_analise = [
    'ValorContratado',
    'ReceitaBrutaMensal',
    'QuantidadeParcelas',
    'TotalContratacoes',
    'TotalContratacoesAtraso'
] + novas_features

matriz = dados[variaveis_analise].corr()

matriz
```
