```python
import os
import re
import pandas as pd
from PyPDF2 import PdfReader

# ---------------------------------------
# CONFIGURAÇÕES
# ---------------------------------------
#SCOPES = ['https://www.googleapis.com/auth/drive.readonly']
#PDF_FOLDER_ID = 'PASTE_HERE_THE_FOLDER_ID'  # <-- ID da pasta no Google Drive

folder_path = "/content/drive/MyDrive/mestrado/artigos/"

terms = [
    "mutual information"
]

# Função para contar ocorrências
def count_terms(text, term):
    pattern = r'\b' + re.escape(term.lower()) + r'\b'
    return len(re.findall(pattern, text.lower()))

# Lista onde ficará o resultado consolidado
results = []

# Percorre todos os arquivos da pasta
for file in os.listdir(folder_path):
    if file.lower().endswith(".pdf"):
        pdf_path = os.path.join(folder_path, file)
        print(f"Lendo PDF: {file}")

        # Extrai texto do PDF
        reader = PdfReader(pdf_path)
        full_text = ""

        for page in reader.pages:
            full_text += page.extract_text() or ""

        # Conta os termos
        counts = {term: count_terms(full_text, term) for term in terms}
        counts["arquivo"] = file

        # Adiciona ao resultado
        results.append(counts)

# Gera DataFrame consolidado
df = pd.DataFrame(results)

# Exporta para Excel no Drive
output_path = "/content/drive/MyDrive/levantamento_temas_drive.xlsx"
df.to_excel(output_path, index=False)

print("Arquivo gerado em:", output_path)
df
```

```python
import os
import re
import pandas as pd
from PyPDF2 import PdfReader

# ---------------------------------------
# CONFIGURAÇÕES
# ---------------------------------------
folder_path = "/content/drive/MyDrive/mestrado/artigos/"  # <-- mesma lógica que já funciona

# Regex para achar anos entre 1900 e 2099
year_pattern = r"\b(19[0-9]{2}|20[0-9]{2})\b"

# Função para identificar o ano mais provável
def extract_year(text):
    years = re.findall(year_pattern, text)
    if not years:
        return None
    # Retorna o ano mais frequente no PDF
    return max(set(years), key=years.count)

# Lista com os resultados
results = []

# Percorre todos os PDFs da pasta
for file in os.listdir(folder_path):
    if file.lower().endswith(".pdf"):
        pdf_path = os.path.join(folder_path, file)
        print(f"Lendo PDF: {file}")

        # Extrai texto
        reader = PdfReader(pdf_path)
        full_text = ""

        for page in reader.pages:
            full_text += page.extract_text() or ""

        # Extrai o ano
        year = extract_year(full_text)

        results.append({
            "arquivo": file,
            "ano_publicacao": year
        })

# Converte para DataFrame
df = pd.DataFrame(results)

# Exporta para Excel
output_path = "/content/drive/MyDrive/levantamento_anos_publicacao.xlsx"
df.to_excel(output_path, index=False)

print("Arquivo gerado em:", output_path)
df
```
